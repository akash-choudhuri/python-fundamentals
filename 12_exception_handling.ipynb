# Python Fundamentals: Exception Handling
Gracefully catch and handle structural errors without crashing your script execution runs.


## 1. Basic Try-Except Isolation
Intercepting standard code runtime failures cleanly.


```python
try:
    division_trap = 10 / 0
except ZeroDivisionError as error:
    print(f'Intercepted zero division hazard: {error}')
```

## 2. Targeting Multiple Errors Specificity
Handling different types of exceptions with distinct blocks.


```python
try:
    string_num = 'abc'
    converted_val = int(string_num)
except ValueError as val_error:
    print(f'ValueError: {val_error}')
except TypeError as type_error:
    print(f'TypeError: {type_error}')
```

## 3. Utilizing Else Blocks
The `else` block runs only if code execution inside the `try` block succeeds without any errors.


```python
try:
    clean_math = 100 / 5
except ZeroDivisionError:
    print('An error occurred.')
else:
    print(f'Success! Computed arithmetic output value: {clean_math}')
```

## 4. Guaranteeing Code Cleanup with Finally
The `finally` block runs no matter what, regardless of whether exceptions were raised or caught.


```python
try:
    print('Executing transaction tasks...')
    faulty_operation = [1, 2][5] # Index out of range
except IndexError as index_err:
    print(f'Index problem identified: {index_err}')
finally:
    print('Finally execution check: Releasing database connection blocks.')
```

## 5. Raising Manual Assertions
Intentionally forcing errors to fire when business logic criteria breaks downstream configurations.


```python
def verify_age(user_age):
    if user_age < 0:
        raise ValueError('Age cannot be evaluated as negative numbers.')
    return f'Validated age status: {user_age}'

try:
    verify_age(-5)
except ValueError as custom_error:
    print(f'Custom Verification Failed: {custom_error}')
```
