# Python Fundamentals: Strings and Methods
Strings are immutable sequences of characters enclosed in quotes.


## 1. Indexing and Slicing
Extract segments using bracket notation syntax `[start:stop:step]`.


```python
phrase = 'Python_Automation'
print('First char:', phrase[0])
print('Last char:', phrase[-1])
print('Slice Python:', phrase[0:6])
print('Reverse string:', phrase[::-1])
```

## 2. Modifying Text Case & Structure
Using string transformation protocols.


```python
sample = '  learn Python programming  '
print('Uppercase:', sample.upper())
print('Stripped Whitespace:', sample.strip())
print('Title Case:', sample.title())
```

## 3. Searching and Replacing
Locating items and swapping data inside strings.


```python
text = 'The quick brown fox jumps over the lazy dog'
print('Contains fox?:', 'fox' in text)
print('Find index of quick:', text.find('quick'))
print('Replace dog:', text.replace('dog', 'cat'))
```

## 4. String Formatting
Using modern f-strings and legacy `.format()` functions.


```python
name, framework = 'Alex', 'Jupyter'
f_string_output = f'Hello {name}, welcome to {framework}.'
format_output = 'Hello {}, welcome to {}.'.format(name, framework)
print(f_string_output)
print(format_output)
```
