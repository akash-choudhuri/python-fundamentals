# Python Fundamentals: Functions and Lambda Expressions
Encapsulate code logic into reusable functions.


## 1. Standard Function Structure
Declaring functions with custom naming conventions and parameters.


```python
def calculate_area(width, height):
    return width * height

result = calculate_area(5, 10)
print('Calculated Area:', result)
```

## 2. Default Parameters
Provide alternative parameter assignments in case fallback paths are needed.


```python
def send_alert(message, severity='Low'):
    return f'[{severity}] Notification: {message}'

print(send_alert('Disk near full', 'Critical'))
print(send_alert('Backup complete'))
```

## 3. Positional and Keyword Arguments
Calling logic configurations by tracking explicit names.


```python
def identify_user(name, occupation):
    print(f'Name: {name} | Job: {occupation}')

identify_user(occupation='Engineer', name='Bob') # Keyword execution
```

## 4. Arbitrary Arguments (*args, **kwargs)
Accepting an arbitrary number of positional or keyword parameters.


```python
def total_sum(*args):
    return sum(args)

def print_meta(**kwargs):
    print(kwargs)

print('Sum:', total_sum(10, 20, 30))
print_meta(app='VSCode', port=8080)
```

## 5. Lambda Expressions
Small, single-expression anonymous functions built on the fly.


```python
square_func = lambda x: x ** 2
multiplier = lambda a, b: a * b
print('Square Lambda:', square_func(8))
print('Multiplier Lambda:', multiplier(3, 7))
```
