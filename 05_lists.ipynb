# Python Fundamentals: Lists
Lists are ordered, mutable collections that allow duplicate values.


## 1. Initialization and Accessing


```python
matrix_list = ['Data', 42, True, [1, 2]]
print('First Item:', matrix_list[0])
print('Nested Item:', matrix_list[3][1])
```

## 2. Adding and Extending Lists
Modifying dimensions with `.append()`, `.insert()`, and `.extend()`.


```python
items = ['apple', 'banana']
items.append('orange')
items.insert(1, 'blueberry')
items.extend(['grape', 'kiwi'])
print('Updated List:', items)
```

## 3. Removing Elements
Using `.pop()`, `.remove()`, and `clear()`.


```python
items.remove('banana') # Deletes item by name
popped_item = items.pop() # Pops off last element
print('Remaining:', items)
print('Popped item:', popped_item)
```

## 4. Sorting and Copying
Be careful with referencing versus direct copying.


```python
nums = [4, 2, 9, 1]
nums.sort()
copied_nums = nums.copy()
print('Sorted items:', nums)
```

## 5. List Comprehension
A concise syntax for creating new lists dynamically.


```python
squares = [x**2 for x in range(6) if x % 2 == 0]
print('Even squares list comprehension:', squares)
```
