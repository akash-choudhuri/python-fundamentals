# Python Fundamentals: Tuples and Sets
Tuples are immutable ordered sequences. Sets are unordered, unique, unindexed collections.


## 1. Tuples: Creation and Immutability


```python
immutable_tuple = ('Linux', 'Windows', 'macOS')
print('Access Tuple Index:', immutable_tuple[1])
print('Length:', len(immutable_tuple))
```

## 2. Tuple Unpacking and Methods
Extracting values into standalone variable allocations safely.


```python
coordinates = (10, 20, 30)
x, y, z = coordinates # Unpacking
print(f'Unpacked fields: x={x}, y={y}, z={z}')
test_tuple = (1, 2, 2, 3)
print('Count occurrences of 2:', test_tuple.count(2))
```

## 3. Sets: Eliminating Duplicates
Sets contain unique elements only.


```python
raw_data = [1, 2, 2, 3, 4, 4, 4, 5]
unique_set = set(raw_data)
print('Set Object (Duplicates automatically removed):', unique_set)
unique_set.add(6)
unique_set.remove(1)
print('Modified Set:', unique_set)
```

## 4. Set Mathematical Operations
Venn diagram interactions: Union, Intersection, and Difference.


```python
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}
print('Union:', set_a.union(set_b))
print('Intersection:', set_a.intersection(set_b))
print('Difference (A - B):', set_a.difference(set_b))
```
