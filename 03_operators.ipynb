# Python Fundamentals: Operators
Operators execute computational logic on values and variables.


## 1. Arithmetic & Floor Operators
Standard arithmetic alongside floor division (`//`) and modulo (`%`).


```python
a, b = 10, 3
print('Addition:', a + b)
print('Standard Division:', a / b)
print('Floor Division (Truncated):', a // b)
print('Modulo (Remainder):', a % b)
print('Exponentiation:', 2 ** 3)
```

## 2. Comparison Operators
Evaluate conditions to return `True` or `False` status.


```python
x, y = 15, 20
print('Equal:', x == y)
print('Not Equal:', x != y)
print('Greater than or equal:', y >= x)
```

## 3. Logical Operators
Used to combine logical assertions: `and`, `or`, and `not`.


```python
cond1 = (5 > 3)
cond2 = (10 < 2)
print('AND evaluation:', cond1 and cond2)
print('OR evaluation:', cond1 or cond2)
print('NOT inversion:', not cond1)
```

## 4. Identity & Membership Operators
Use `is` to check object memory identification, and `in` for data sequences.


```python
list_a = [1, 2, 3]
list_b = list_a
list_c = [1, 2, 3]
print('Identity Check (Same object):', list_a is list_b)
print('Identity Check (Copy object):', list_a is list_c)
print('Membership Check:', 2 in list_a)
```
