# Python Fundamentals: Data Types and Type Casting
Data types determine what operations can be performed on specific values.


## 1. Numeric Types
Integers, floats, and complex numbers.


```python
integer_num = 42
float_num = 3.1415
complex_num = 2 + 3j
print(type(integer_num), type(float_num), type(complex_num))
```

## 2. Structural & Boolean Data Types
Booleans store truth values, sequences store structural configurations.


```python
is_active = True
none_val = None
print('Boolean:', type(is_active))
print('NoneType:', type(none_val))
```

## 3. Implicit Type Conversion
Python converts data types automatically when safe (e.g., mixing integer and float).


```python
num_int = 10
num_flo = 5.5
result = num_int + num_flo
print('Result value:', result, '| Type:', type(result))
```

## 4. Explicit Type Conversion (Casting)
Using built-in functions like `int()`, `float()`, or `str()` manually.


```python
age_str = '25'
age_int = int(age_str)
pi_int = int(3.99) # Drops everything after the decimal point
str_back = str(pi_int)
print(age_int + 5, pi_int, type(str_back))
```
