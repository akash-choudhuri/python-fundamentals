# Python Fundamentals: Variables and Constants
Variables are named locations used to store data in memory.


## 1. Variable Assignment
Python is dynamically typed. You don't need to state the variable type explicitly.


```python
x = 5
y = 'Hello'
print('x:', x, '| type:', type(x))
print('y:', y, '| type:', type(y))
```

## 2. Dynamic Re-assignment
A variable can change its data type at any point during script execution.


```python
var = 100
print('Initial:', var, type(var))
var = 'Now I am a string'
print('Changed:', var, type(var))
```

## 3. Multiple Assignments
You can assign multiple variables simultaneously or assign the same value to many variables.


```python
a, b, c = 1, 2.5, 'Three'
print(a, b, c)
x = y = z = 0
print(x, y, z)
```

## 4. Constants and Naming Rules
Constants use UPPERCASE names by convention. Variables cannot start with numbers or contain special symbols except `_`.


```python
PI = 3.14159
MAX_CONNECTIONS = 100
user_age = 25  # snake_case is standard
print(f'PI: {PI}, Max Connections: {MAX_CONNECTIONS}')
```
