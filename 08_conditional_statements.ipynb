# Python Fundamentals: Conditional Statements
Control code execution pathways based on Boolean decisions.


## 1. Basic If-Else Blocks


```python
temperature = 28
if temperature > 30:
    print('It is hot outside.')
else:
    print('Weather is mild or cold.')
```

## 2. Multi-Condition Elif Chains
Evaluating conditions sequentially from top to bottom.


```python
score = 78
if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
else:
    grade = 'F'
print(f'Assigned Grade: {grade}')
```

## 3. Nested Conditionals
Placing conditional blocks inside other conditional blocks.


```python
num = -10
if num >= 0:
    if num == 0:
        print('Zero')
    else:
        print('Positive Number')
else:
        print('Negative Number')
```

## 4. Inline Ternary Operators
A shorthand way to evaluate an expression on a single line.


```python
age = 20
status = 'Adult' if age >= 18 else 'Minor'
print('Status:', status)
```
