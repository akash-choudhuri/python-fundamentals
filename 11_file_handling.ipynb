# Python Fundamentals: File Handling
Interact with the host filesystem to read and write persistent data structures safely.


## 1. Writing to Text Files
Use the context manager `with` block to ensure system assets close securely.


```python
with open('dataset.txt', 'w') as writer:
    writer.write('Row_1: Configuration Active\nRow_2: System Healthy\n')
print('File payload written successfully.')
```

## 2. Appending Records
Adding log entries onto existing files using `a` mode without overwriting them.


```python
with open('dataset.txt', 'a') as appender:
    appender.write('Row_3: Script Completed Execution\n')
print('Appended details lines safely.')
```

## 3. Reading Entire Files
Consuming whole files into a single string allocation.


```python
with open('dataset.txt', 'r') as reader:
    payload = reader.read()
print('--- Complete Read File Payload ---')
print(payload)
```

## 4. Reading Line by Line
Looping through file contexts tracking elements one array lane at a time.


```python
with open('dataset.txt', 'r') as target_file:
    for index, line in enumerate(target_file, start=1):
        print(f'Line tracking index {index}: {line.strip()}')
```

## 5. Clean-Up File Operations
Removing files using the standard `os` module tools.


```python
import os
if os.path.exists('dataset.txt'):
    os.remove('dataset.txt')
    print('Cleaned up sandbox space: dataset.txt removed.')
else:
    print('Target file does not exist.')
```
