# Python Fundamentals: Loops
Automate repeated code executions smoothly using loops.


## 1. For Loops with Range Configurations
Iterate over an explicit sequence of numbers using `range(start, stop, step)`.


```python
for i in range(1, 10, 2):
    print(f'Odd count sequence value: {i}')
```

## 2. Iterating Sequences
Traversing elements directly in data lists or character patterns directly.


```python
languages = ['Python', 'Go', 'Rust']
for lang in languages:
    print(f'Engineered Language entry: {lang}')
```

## 3. While Loops
Execute block code continually as long as a condition remains True.


```python
countdown = 3
while countdown > 0:
    print(f'T-Minus: {countdown}')
    countdown -= 1
print('Blast off!')
```

## 4. Break Statement
Exit the entire loop immediately, bypassing remaining elements.


```python
for n in range(1, 100):
    if n == 5:
        print(f'Broke out of loop early at n={n}')
        break
    print('Loop index:', n)
```

## 5. Continue Statement
Skip the rest of the current iteration and jump straight to the next loop pass.


```python
for item in range(1, 6):
    if item == 3:
        print('Skipping item index 3 via continue statement')
        continue
    print('Processing item:', item)
```
