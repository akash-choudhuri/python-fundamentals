# Python Fundamentals: Dictionaries
Dictionaries store data in mutable key:value pairs mapping schemas.


## 1. Creating and Querying Keys


```python
user_profile = {'username': 'coder123', 'clearance': 'Admin', 'id': 9901}
print(user_profile['username'])
```

## 2. Safe Value Retrievals
Using `.get()` avoids runtime KeyErrors when a key does not exist.


```python
# print(user_profile['email']) # Throws KeyError
email_lookup = user_profile.get('email', 'No Email Registered')
print('Safe Fetch Result:', email_lookup)
```

## 3. Modifications and Updates
Adding, updating, and removing mappings.


```python
user_profile['email'] = 'test@domain.com' # Assign new key
user_profile.update({'clearance': 'SuperAdmin', 'status': 'Active'})
del user_profile['id']
print('Modified Profile:', user_profile)
```

## 4. Extraction Views
Grabbing subsets via `.keys()`, `.values()`, and `.items()` structural tracks.


```python
print('Keys View:', list(user_profile.keys()))
print('Values View:', list(user_profile.values()))
```

## 5. Iterating Dictionaries
Iterating through your keys and values simultaneously with `.items()`.


```python
for key, val in user_profile.items():
    print(f'Setting: Key={key} | Value={val}')
```
